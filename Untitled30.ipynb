# Fraud Class Distribution

In this section, we visualize the distribution of normal and fraudulent transactions in the dataset.

```python
print(df['Class'].value_counts())

plt.figure(figsize=(6,4))
sns.countplot(x='Class', data=df)

plt.title('Class Distribution')
plt.xlabel('Class')
plt.ylabel('Count')

plt.show()
```

# Feature Scaling and Data Preparation

In this section, we scale the numerical features using StandardScaler to prepare the dataset for machine learning and deep learning algorithms.

```python

X = df.drop('Class', axis=1)
y = df['Class']


scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Feature scaling completed!")
print("Scaled data shape:", X_scaled.shape)
```

# Dimensionality Reduction using PCA

In this section, we apply Principal Component Analysis (PCA) to reduce the dimensionality of the dataset while preserving the most important information.

```python

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

print("Original Shape:", X_scaled.shape)
print("Reduced Shape:", X_pca.shape)
```

PCA VISULAISATION

```python

plt.figure(figsize=(8,6))

plt.scatter(X_pca[:,0], X_pca[:,1], c=y, cmap='coolwarm', s=2)

plt.title('PCA Visualization')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')

plt.show()
```

# Data Visualization using t-SNE

In this section, we apply t-SNE to visualize the high-dimensional data in a two-dimensional space and observe hidden patterns and anomalies.

```python

sample_size = 5000

X_sample = X_scaled[:sample_size]
y_sample = y[:sample_size]
```

```python

tsne = TSNE(n_components=2, random_state=42)

X_tsne = tsne.fit_transform(X_sample)

print("t-SNE completed!")
```

```python
# t-SNE Visualization
plt.figure(figsize=(8,6))

plt.scatter(X_tsne[:,0], X_tsne[:,1], c=y_sample, cmap='coolwarm', s=5)

plt.title('t-SNE Visualization')
plt.xlabel('t-SNE Component 1')
plt.ylabel('t-SNE Component 2')

plt.show()
```
