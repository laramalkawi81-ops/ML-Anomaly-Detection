# Feature Scaling and Data Preparation

In this section, we scale the numerical features using StandardScaler to prepare the dataset for machine learning and deep learning algorithms.

```python

X = df.drop('Class', axis=1)
y = df['Class']


scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Feature scaling completed!")
print("Scaled data shape:", X_scaled.shape)
```

# Dimensionality Reduction using PCA

In this section, we apply Principal Component Analysis (PCA) to reduce the dimensionality of the dataset while preserving the most important information.

```python

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

print("Original Shape:", X_scaled.shape)
print("Reduced Shape:", X_pca.shape)
```

PCA VISULAISATION

```python

plt.figure(figsize=(8,6))

plt.scatter(X_pca[:,0], X_pca[:,1], c=y, cmap='coolwarm', s=2)

plt.title('PCA Visualization')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')

plt.show()
```

# Data Visualization using t-SNE

In this section, we apply t-SNE to visualize the high-dimensional data in a two-dimensional space and observe hidden patterns and anomalies.

```python

sample_size = 5000

X_sample = X_scaled[:sample_size]
y_sample = y[:sample_size]
```

```python

tsne = TSNE(n_components=2, random_state=42)

X_tsne = tsne.fit_transform(X_sample)

print("t-SNE completed!")
```

```python
# t-SNE Visualization
plt.figure(figsize=(8,6))

plt.scatter(X_tsne[:,0], X_tsne[:,1], c=y_sample, cmap='coolwarm', s=5)

plt.title('t-SNE Visualization')
plt.xlabel('t-SNE Component 1')
plt.ylabel('t-SNE Component 2')

plt.show()
```

# Clustering using DBSCAN

In this section, we apply the DBSCAN algorithm to identify clusters in the dataset and detect noise points (outliers).

```python
sample_size = 10000

X_sample = X_scaled[:sample_size]
```

```python
dbscan = DBSCAN(eps=10, min_samples=5)
db_labels = dbscan.fit_predict(X_sample)
```

```python
plt.scatter(X_pca[:sample_size,0], X_pca[:sample_size,1], c=db_labels, s=2)
```

# Anomaly Detection using One-Class SVM

In this section, we apply One-Class SVM to detect anomalies in the dataset.

```python

sample_size = 10000
X_sample = X_scaled[:sample_size]


ocsvm = OneClassSVM(kernel='rbf', gamma='auto', nu=0.01)

y_pred = ocsvm.fit_predict(X_sample)

print("One-Class SVM completed!")
```

```python


unique, counts = np.unique(y_pred, return_counts=True)

print(dict(zip(unique, counts)))
```

```python
plt.figure(figsize=(8,6))

plt.scatter(X_pca[:sample_size,0], X_pca[:sample_size,1],
            c=y_pred, cmap='coolwarm', s=2)

plt.title('One-Class SVM Anomaly Detection')
plt.xlabel('PCA Component 1')
plt.ylabel('PCA Component 2')

plt.show()
```

# Deep Learning using Autoencoder

In this section, we build an autoencoder neural network to detect anomalies in the dataset.

```python

input_dim = X_sample.shape[1]
input_layer = Input(shape=(input_dim,))


encoded = Dense(16, activation='relu')(input_layer)
encoded = Dense(8, activation='relu')(encoded)

decoded = Dense(16, activation='relu')(encoded)
decoded = Dense(input_dim, activation='sigmoid')(decoded)

autoencoder = Model(inputs=input_layer, outputs=decoded)

autoencoder.compile(optimizer='adam', loss='mse')

autoencoder.summary()
```

```python
history = autoencoder.fit(
    X_sample, X_sample,
    epochs=10,
    batch_size=256,
    validation_split=0.2,
    shuffle=True
)
```

```python
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])

plt.title('Autoencoder Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')

plt.show()
```

```python
reconstructions = autoencoder.predict(X_sample)


mse = np.mean(np.power(X_sample - reconstructions, 2), axis=1)


threshold = np.percentile(mse, 95)


ae_labels = (mse > threshold).astype(int)

print("Anomalies detected:", np.sum(ae_labels))
```

```python
plt.scatter(X_pca[:sample_size,0], X_pca[:sample_size,1],
            c=ae_labels, cmap='coolwarm', s=2)

plt.title('Autoencoder Anomaly Detection')
plt.xlabel('PCA Component 1')
plt.ylabel('PCA Component 2')

plt.show()
```

# Model Evaluation

Different models are evaluated using appropriate metrics such as precision, recall, and F1-score to compare their performance in anomaly detection.

```python
from sklearn.metrics import classification_report, confusion_matrix

y_true = y[:sample_size]


y_pred_svm = np.where(y_pred == -1, 1, 0)

print(confusion_matrix(y_true, y_pred_svm))
print(classification_report(y_true, y_pred_svm))
```


Autoencoder Metrics

```python
y_pred_ae = ae_labels

print(confusion_matrix(y_true, y_pred_ae))
print(classification_report(y_true, y_pred_ae))
```
